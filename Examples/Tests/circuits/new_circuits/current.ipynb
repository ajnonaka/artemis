# Iloop current vs time

Reads 4 text files (skipping the first line). Each remaining line has 5 columns.

Builds a 2D array:
- `arr2d[:,0]` (time) = 2nd column of `Iloop_top.txt`
- `arr2d[:,1]` (current) = (3rd col of top) − (3rd col of bottom) + (5th col of left) − (5th col of right)

Then plots current vs time.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({"figure.dpi": 130})
```

```python
# ---- File paths ----
top_path = "Iloop_top.txt"
bottom_path = "Iloop_bottom.txt"
left_path = "Iloop_left.txt"
right_path = "Iloop_right.txt"
```

```python
# ---- Read files (skip first line; expect 5 columns) ----
def read_5col_file(path):
    data = np.loadtxt(path, skiprows=1)
    if data.ndim != 2 or data.shape[1] != 5:
        raise ValueError(f"Expected a 2D array with 5 columns in {path}, got shape {getattr(data, 'shape', None)}")
    return data

top = read_5col_file(top_path)
bottom = read_5col_file(bottom_path)
left = read_5col_file(left_path)
right = read_5col_file(right_path)

# sanity check: same number of rows
n = top.shape[0]
if not (bottom.shape[0] == left.shape[0] == right.shape[0] == n):
    raise ValueError(
        f"Row counts do not match: top={top.shape[0]}, bottom={bottom.shape[0]}, left={left.shape[0]}, right={right.shape[0]}"
    )

top[:5]
```

```python
# ---- Build the 2D array ----
# 0-based indices:
# 2nd column -> 1
# 3rd column -> 2
# 4th column -> 3
# 5th column -> 4

time = top[:, 1]
current = top[:, 2] - bottom[:, 2] + left[:, 4] - right[:, 4]

arr2d = np.column_stack([time, current])

arr2d[:5]
```

```python
# ---- Plot ----
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(arr2d[:, 0], arr2d[:, 1], lw=1.5)
ax.set_xlabel("Time")
ax.set_ylabel("Current")
ax.set_title("Iloop current vs time")
ax.grid(True, alpha=0.3)
plt.show()
```
